1. Explain pandas and its role in Data Analysis and Machine Learning.

# ans->
Pandas is a Python library used to work with data in a simple and organized way. It helps us read, clean, analyze, and manage large amounts of data. Pandas mainly uses two data structures: Series (single column) and DataFrame (table with rows and columns).

### Role of Pandas in Data Analysis
Used to read data from CSV, Excel, and other files.
Helps in cleaning data by removing errors and handling missing values.
Allows filtering, sorting, and selecting specific data.
Makes data analysis easier by providing statistical functions.
Helps prepare data for visualization.
### Role of Pandas in Machine Learning
Used to prepare datasets before training machine learning models.
Helps handle missing values and incorrect data.
Used for feature selection and data transformation.
Organizes data into input (X) and output (y) variables.
Makes data ready for model training and testing.

2. Differentiate between:
● Series and DataFrame
● loc[] and iloc[] functions in Pandas

#### (a) Series vs DataFrame
| Feature   | Series            | DataFrame               |
| --------- | ----------------- | ----------------------- |
| Dimension | 1-Dimensional     | 2-Dimensional           |
| Structure | Single Column     | Multiple Columns        |
| Index     | Has Index         | Has Rows and Columns    |
| Example   | Marks of students | Complete student record |
| Creation  | pd.Series()       | pd.DataFrame()          |

#### (b) loc[] vs iloc[]
| Feature   | loc[]            | iloc[]         |
| --------- | ---------------- | -------------- |
| Selection | Label Based      | Position Based |
| Uses      | Row/Column Names | Integer Index  |
| End Index | Included         | Excluded       |


3. Create a Pandas DataFrame using the dataset below:
Name Age Marks
Amit 20 85
Neha 21 90
Rahul 19 78
Sneha 22 88
a) Tasks:
● Display first 2 rows
● Display only Name and Marks columns
● Add a new column named “Result”
● Display students with marks greater than 80
b) Expected Concepts:
● pd.DataFrame()
● head()
● Column Selection
● Conditional Filtering

```python
import pandas as pd
data = {
    'Name':['Amit','Neha','Rahul','Sneha'],
    'Age':[20,21,19,22],
    'Marks':[85,90,78,88]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

#### (a)Tasks: 
● Display first 2 rows
● Display only Name and Marks columns
● Add a new column named “Result”
● Display students with marks greater than 80

```python
print("Display First 2 Rows")
print(df.head(2))

print("Display Only Name and Marks Columns")
print(df[['Name','Marks']])

print("Add New Column Result")
df['Result'] = ['Pass','Pass','Pass','Pass']
print(df,"\n")

print("Display Students with Marks Greater Than 80")
print(df[df['Marks'] > 80])
```

4. Create a DataFrame using the dataset below:
Name Age Salary
A 25 30000
B NaN 40000
C 27 NaN
D 24 35000
a) Tasks:
● Identify missing values
● Fill missing Age with average age
● Fill missing Salary with median salary
● Display cleaned dataset
b) Expected Concepts:
● isnull()
● fillna()
● mean()
● median()

```python
import pandas as pd
import numpy as np

data = {
    'Name':['A','B','C','D'],
    'Age':[25,np.nan,27,24],
    'Salary':[30000,40000,np.nan,35000]
}
print("\nCreate DataFrame")
df = pd.DataFrame(data)
print(df)

print("\nIdentify Missing Values")
print(df.isnull())
print(df.isnull().sum())

print("\nFill Missing Age with Average Age")
avg_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(avg_age)

print("\nFill Missing Age with Average Age")
median_salary = df['Salary'].median()
df['Salary']=df['Salary'].fillna(median_salary)

print("\nDisplay Cleaned Dataset")
print(df)
```
